```python
from IPython.display import Audio
Audio(AUDIO_PATH)
```

```python
def plot_waveform(signal):
    plt.figure(figsize=(10, 4))
    plt.plot(signal)
    plt.title("Audio Waveform")
    plt.xlabel("Time (samples)")
    plt.ylabel("Amplitude")
    plt.tight_layout()
    plt.show()
```
